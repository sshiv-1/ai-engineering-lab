```python
import pandas as pd
import numpy as np
```

```python
df=pd.read_csv('https://raw.githubusercontent.com/Slimani-CE/diebetes-prediction/refs/heads/main/dataset/diabetes.csv')
df.head()
```

```python
df.corr()['Outcome']
```

```python
X=df.drop(columns=['Outcome'])
y=df['Outcome']
```

```python
from sklearn.preprocessing import StandardScaler
scaler=StandardScaler()
X_scaled=scaler.fit_transform(X)

X_scaled
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(X_scaled,y,random_state=42,shuffle=True,train_size=0.2)
```

```python
import tensorflow as tf
from tensorflow.keras.optimizers import Adam
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Input, Dense, Dropout
from tensorflow.keras.regularizers import l2
```

```python
model=Sequential([
    Input((8,)),
    Dense(128,activation='relu',kernel_regularizer=l2(0.05)),
    Dropout(0.3),
    Dense(32,activation='relu',kernel_regularizer=l2(0.05)),
    Dropout(0.3),
    Dense(1,activation='sigmoid')
])
adam=Adam(learning_rate=0.01)
model.compile(optimizer=adam,loss='binary_crossentropy',metrics=['accuracy'])
```

```python
model.summary()
```

```python
history=model.fit(X_train,y_train,epochs=250,batch_size=32,validation_data=(X_test,y_test))
```

```python
import keras_tuner as kt
```

```python
def build_model(hp):
    model=Sequential([
        Input((8,)),
        Dense(128,activation='relu'),
        Dense(1,activation='sigmoid')
    ])

    model.compile(optimizer=hp.Choice('optimizer_select',['adam','rmsprop','adagrad','nadam','adadelta','sgd']),loss='binary_crossentropy',metrics=['accuracy'] )
    return model
```

```python
tuner=kt.RandomSearch(build_model,objective='val_accuracy',max_trials=5)
```

```python
tuner.search(X_train,y_train,epochs=50,validation_data=(X_test,y_test))
```

```python
tuner.get_best_hyperparameters()[0].values
```

```python
model=tuner.get_best_models(num_models=1)[0]
model.summary()
```

```python
def build_model(hp):
    units= hp.Int('units',8,128,step=8)
    model=Sequential([
        Input(shape=(8,)),
        Dense(units,activation='relu'),
        Dense(1,activation='sigmoid')
    ])
    model.compile(optimizer='rmsprop',loss='binary_crossentropy',metrics=['accuracy'] )

    return model
```

```python
tuner1 = kt.RandomSearch(
    build_model,
    objective='val_accuracy',
    max_trials=5,
    directory='kt_dir',
    project_name='simple_nn'
)
```

```python
tuner1.search(X_train,y_train,epochs=50,validation_data=(X_test,y_test))
```

```python
tuner1.get_best_hyperparameters()[0].values
```

```python
model=tuner1.get_best_models(num_models=1)[0]
model.summary()
```

```python
def build_model(hp):
    model=Sequential()
    model.add(Input(shape=(8,)))
    for i in range(hp.Int('units',min_value=1,max_value=10,)):
        model.add(Dense(32,activation='relu'))

    model.add(Dense(1,activation='sigmoid'))
    model.compile(optimizer='rmsprop',loss='binary_crossentropy',metrics=['accuracy'] )
    return model
```

```python
tuner1 = kt.RandomSearch(
    build_model,
    objective='val_accuracy',
    max_trials=5,
    directory='kt_dir',
    project_name='no_of_layers'
)
```

```python
tuner1.search(X_train,y_train,epochs=50,validation_data=(X_test,y_test))
```

```python
tuner1.get_best_hyperparameters()[0].values
```

```python
model=tuner1.get_best_models(num_models=1)[0]
model.summary()
```
