```python
from keras.applications.vgg16 import VGG16,preprocess_input
vgg=VGG16()
```

```python
vgg.summary()
```

```python
vgg.layers
```

```python
for i in range(len(vgg.layers)):
    if 'conv' in vgg.layers[i].name:
        filters,bias=vgg.layers[i].get_weights()
        print(f"layer no.{i}: {vgg.layers[i].name} ,{filters.shape}")
```

```python
# retreive w from first conv layer
filters,bias=vgg.layers[1].get_weights()
print(f"weights of second layer: {filters}")
```

```python
# normalisation
f_min,f_max=filters.min(),filters.max()
filters=(filters-f_min)/(f_max-f_min)
filters
```

```python
from matplotlib import pyplot
n_filters = 6
ix = 1
fig = pyplot.figure(figsize=(15,10))

for i in range(n_filters):
    # get the filters
    f = filters[:, :, :, i]
    for j in range(3):
        # subplot for 6 filters and 3 channels
        pyplot.subplot(n_filters, 3, ix)
        pyplot.imshow(f[:, :, j], cmap='gray')
        ix += 1

# plot the filters
pyplot.show()
```

```python
from keras.models import Model
import tensorflow as tf
model=Model(inputs=vgg.inputs,
            outputs=vgg.layers[1].output)
```

```python
from keras.utils import load_img,img_to_array
img = load_img(
    "C:/Coding/ML_DL/Data-Aurgumentation/train/cat.jpg",
    target_size=(224, 224)
)
```

```python
img=img_to_array(img)
img=tf.expand_dims(img,axis=0)
img=preprocess_input(img)
```

```python
features=model.predict(img)
```

```python
fig=pyplot.figure(figsize=(20,15))
for i in range(1,features.shape[3]+1):
    pyplot.subplot(8,8,i)
    pyplot.imshow(features[0,:,:,i-1],cmap='gray')
pyplot.show()
```

```python
layer_index = [ 2, 5, 9, 13, 17]
model2=VGG16()
outputs = [model2.layers[i].output for i in layer_index]

model3 = Model( inputs= model2.inputs, outputs = outputs)
```

```python
feature_map=model3.predict(img)
```

```python
for layer_idx, fm in zip(layer_index, feature_map):

    fig = pyplot.figure(figsize=(20,15))
    fig.suptitle(f"Layer_{layer_idx}", fontsize=20)

    n_filters = fm.shape[3]

    for j in range(1, min(n_filters, 64) + 1):
        pyplot.subplot(8, 8, j)
        pyplot.imshow(fm[0, :, :, j-1], cmap='gray')
        pyplot.axis('off')

    pyplot.show()
```
